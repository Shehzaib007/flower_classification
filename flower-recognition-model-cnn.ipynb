```python
#installing and importing libraries
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import image
import os
```

```python
# 1. Load the trained model from Kaggle Models
model_path = "/kaggle/input/flowers-recognition-cnn-training/other/default/1/flowers_cnn.h5"
model = tf.keras.models.load_model(model_path)

# 2. Define class names (same order used during training)
class_names = ['daisy', 'dandelion', 'rose', 'sunflower', 'tulip']

# 3. Prediction function for single image
def predict_flower(img_path):
    # Load and preprocess image
    img = image.load_img(img_path, target_size=(128, 128))
    img_array = np.expand_dims(image.img_to_array(img) / 255.0, axis=0)
    
    # Predict probabilities
    preds = model.predict(img_array)
    predicted_index = np.argmax(preds)
    confidence = np.max(preds)
    
    return class_names[predicted_index], confidence
```

```python
# 4. Example usage on one image (replace path with any image)
test_image_path = "/kaggle/input/flowers-recognition/flowers/sunflower/12471290635_1f9e3aae16_n.jpg"
flower, conf = predict_flower(test_image_path)
print(f"Predicted Flower: {flower} (Confidence: {conf:.2f})")

# 5. Optional: Predict multiple images in a folder
def predict_folder(folder_path, limit=5):
    results = []
    count = 0
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith(('.jpg', '.png', '.jpeg')):
                full_path = os.path.join(root, file)
                flower, conf = predict_flower(full_path)
                results.append((file, flower, conf))
                count += 1
                if count >= limit:
                    break
        if count >= limit:
            break
    return results
```
